```python
data=[
    ("zaid",66,15,"India"),
    ("Shivaraj",78,15,"India"),
    ("Jhon",66,50,"UK"),
    ("Sara",97,89,"usa")
]
column=["Name","Age","Expirence","Country"]
```

```python
df=spark.createDataFrame(data , column)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
df.select("Name","Age").show()
```

```python
df.filter(df.Age>66).show()
```
